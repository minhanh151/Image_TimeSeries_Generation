```python
import pandas as pd
import matplotlib.pyplot as plt
max_rows = 24  # limit number of rows
exclude_columns = ['sample_id', 'Volume', 'Adj_Close']

# Load CSV file
csv_file = 'data/stock_data.csv'  # replace with your actual file name
# csv_file = 'data/data.csv'  # replace with your actual file name
df = pd.read_csv(csv_file)

# Limit rows
df = df.tail(max_rows)

# Plot each column as a line
plt.figure(figsize=(12, 6))
for column in df.columns:
   if column not in exclude_columns:
       plt.plot(df.index, df[column], label=column)

# plt.yscale('log')  # Set Y-axis to logarithmic
plt.title('Line Plot per Column for real data')
plt.xlabel('Index')
plt.ylabel('Value')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
